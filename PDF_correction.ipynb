```python
#загрузка библиотек
import pymupdf
import fitz
import pandas as pd
import numpy as np
```

```python
#считывание пдф файла
filename = 'VDMF-11_vild.pdf'
doc = pymupdf.open(filename)
```

```python
#считывание файла эксель
data_to_change=pd.read_excel('data_new_changed.xlsx')
```

```python
#подготовка эксель файла
data_to_change['New area'] = pd.to_numeric(data_to_change['New area'], errors='coerce')
data_to_change['New area']=data_to_change['New area'].round(0)
data_to_change['New conc'] = pd.to_numeric(data_to_change['New conc'], errors='coerce')
data_to_change['New conc']=data_to_change['New conc'].round(2)
```

```python
#перевод пропущеных значений в текстовый формат
data_to_change.fillna('', inplace=True)
```

```python
# изменение площадей
i=0
for new_value in data_to_change['New area'].values:
    if isinstance(new_value, str):
        new_text=new_value

    else:
        new_text=int(new_value)

    raw_text=data_to_change['Old area'].values[i]
    raw_text=raw_text.astype(str)
    new_text=str(new_text)
    page = doc[5]
    hits = page.search_for(raw_text)  # list of rectangles where to replace
    new_fontsize = 20  # Новый размер шрифта
    new_width = -0.1  # Новая ширина прямоугольника
    new_width_2 = -14  # Новая ширина прямоугольника
    new_height = 2.5 # Новая высота прямоугольника
    for rect in hits:
      x1, y1, x2, y2 = rect
      new_x2 = x2 + new_width
      new_y2 = y2 + new_height
      new_x1 = x1 + new_width_2
      new_rect = fitz.Rect(new_x1, y1, new_x2, new_y2)
    for rect in hits:
      page.add_redact_annot(new_rect, new_text,
                              fontname=page.get_fonts()[1][4],fontsize=new_fontsize,
                              align=pymupdf.TEXT_ALIGN_RIGHT)
    i+=1
```

```python
# изменение концентраций
i=0
for new_value in data_to_change['New conc'].values:
    if isinstance(new_value, str):
        new_text = new_value
    else:
        new_text = str(new_value)

    raw_text=data_to_change['Old conc'].values[i]
    raw_text=str(raw_text)
    #new_text=str(new_text)
    new_text=new_text
    page = doc[5]
    if raw_text!='':
      hits = page.search_for(raw_text)  # list of rectangles where to replace
      new_fontsize = 20  # Новый размер шрифта
      new_width = -0.1  # Новая ширина прямоугольника
      new_width_2 = -14  # Новая ширина прямоугольника
      new_height = 2.5 # Новая высота прямоугольника

    for rect in hits:
      x1, y1, x2, y2 = rect
      new_x2 = x2 + new_width
      new_y2 = y2 + new_height
      new_x1 = x1 + new_width_2
      new_rect = fitz.Rect(new_x1, y1, new_x2, new_y2)
    for rect in hits:
      page.add_redact_annot(new_rect, new_text,
                              fontname=page.get_fonts()[1][4],fontsize=new_fontsize,
                              align=pymupdf.TEXT_ALIGN_RIGHT)
      i+=1
```

```python
# запись файла
page.apply_redactions()
doc.save("replaced_1.pdf")#, garbage=3, deflate=True)
```
